# 02 — Turnout and party-list trends

This notebook answers the first two version-1 questions using official parliamentary-election data. Party labels are shown as published by NVI; a party/coalition crosswalk will be added before interpreting long-term ideological trends.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA = PROJECT_ROOT / 'data' / 'processed'

turnout = pd.read_csv(DATA / 'national_turnout_1990_2022.csv')
party_results = pd.read_csv(DATA / 'national_party_list_results_1990_2026.csv')
turnout
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(turnout['election_year'], turnout['turnout_rate'] * 100, marker='o', color='#0B3C5D', linewidth=2)
ax.set(title='Hungarian parliamentary-election turnout', xlabel='Election year', ylabel='Turnout (%)', ylim=(50, 75))
ax.grid(axis='y', alpha=0.25)
for _, row in turnout.iterrows():
    ax.annotate(f"{row['turnout_rate']:.1%}", (row['election_year'], row['turnout_rate'] * 100), xytext=(0, 8), textcoords='offset points', ha='center')
plt.show()

print('Lowest turnout:', turnout.loc[turnout.turnout_rate.idxmin(), ['election_year', 'turnout_rate']].to_dict())
print('Highest turnout:', turnout.loc[turnout.turnout_rate.idxmax(), ['election_year', 'turnout_rate']].to_dict())
```
